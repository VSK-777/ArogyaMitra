```
from pyngrok import ngrok
import nest_asyncio
import uvicorn
import getpass

print("Ngrok now requires an auth token to run.")
print("1. Sign up / Log in to ngrok: https://dashboard.ngrok.com/signup")
print("2. Copy your Authtoken from: https://dashboard.ngrok.com/get-started/your-authtoken")
print("3. Paste it below:\n")
authtoken = getpass.getpass("Enter your ngrok authtoken: ")
ngrok.set_auth_token(authtoken)

public_url = ngrok.connect(8000).public_url
print("\n\n=========================================")
print(f"API is publicly accessible at: {public_url}")
print(f"Update your backend properties to use: {public_url}/summarize")
print("=========================================\n\n")

nest_asyncio.apply()
uvicorn.run("main:app", host="0.0.0.0", port=8000)
```
